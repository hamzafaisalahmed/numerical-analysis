```python
# y = x**2, 1 <= x <= 10

x_list = [i for i in range(1,11)]
y_list = [x**2 for x in x_list]

print(x_list)
print(y_list)
```

```python
def f(x):
    return x**3
```

```python
# y = x**2, 1 <= x <= 10

x_list = [i for i in range(1,11)]
y_list = [f(x) for x in x_list]

print(x_list)
print(y_list)
```
